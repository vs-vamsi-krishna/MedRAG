```python
import pandas as pd
from langchain_core.documents import Document

data_path = "../data/raw"

# Load only the 3 tables needed for our RAG data
applications = pd.read_csv(
    f"{data_path}/Applications.txt",
    sep="\t",
    dtype=str,
    encoding="latin1"
)

products = pd.read_csv(
    f"{data_path}/Products.txt",
    sep="\t",
    dtype=str,
    encoding="latin1"
)

marketing_status = pd.read_csv(
    f"{data_path}/MarketingStatus.txt",
    sep="\t",
    dtype=str,
    encoding="latin1"
)

marketing_lookup = pd.read_csv(
    f"{data_path}/MarketingStatus_Lookup.txt",
    sep="\t",
    dtype=str,
    encoding="latin1"
)

# Join
app_products = pd.merge(
    applications,
    products,
    on="ApplNo",
    how="left"
)

app_products_status = pd.merge(
    app_products,
    marketing_status,
    on=["ApplNo", "ProductNo"],
    how="left"
)

app_products_final = pd.merge(
    app_products_status,
    marketing_lookup,
    on="MarketingStatusID",
    how="left"
)

# Select RAG fields
rag_data = app_products_final[
    [
        "ApplNo",
        "ApplType",
        "SponsorName",
        "ProductNo",
        "Form",
        "Strength",
        "DrugName",
        "ActiveIngredient",
        "MarketingStatusDescription"
    ]
].copy()

rag_data = rag_data.fillna("Not available")

# Create document text
def create_document(row):
    return f"""
Application Number: {row['ApplNo']}
Application Type: {row['ApplType']}
Sponsor: {row['SponsorName']}

Product Number: {row['ProductNo']}
Drug Name: {row['DrugName']}
Active Ingredient: {row['ActiveIngredient']}
Strength: {row['Strength']}
Dosage Form: {row['Form']}
Marketing Status: {row['MarketingStatusDescription']}
""".strip()

rag_data["document_text"] = rag_data.apply(create_document, axis=1)

# Create LangChain Documents
documents = [
    Document(
        page_content=row["document_text"],
        metadata={
            "ApplNo": row["ApplNo"],
            "DrugName": row["DrugName"],
            "ProductNo": row["ProductNo"]
        }
    )
    for _, row in rag_data.iterrows()
]

print("RAG records:", len(rag_data))
print("Documents:", len(documents))
```

### Retrieval process ###

```python
def retrieve_documents(query, k=5):
    
    # Exact drug-name search
    exact_results = vectorstore.get(
        where={"DrugName": query.upper()}
    )

    if exact_results["documents"]:
        return exact_results["documents"][:k]

    # Fallback to semantic search
    results = vectorstore.similarity_search(query, k=k)

    return [doc.page_content for doc in results]
```

```python
import re

def retrieve_documents(query, k=5):

    # Clean punctuation from the query
    words = re.findall(r"\b[\w-]+\b", query.upper())

    # Try exact metadata search for each word
    for word in words:
        exact_results = vectorstore.get(
            where={"DrugName": word}
        )

        if exact_results["documents"]:
            return exact_results["documents"][:k]

    # Fallback to vector search
    results = vectorstore.similarity_search(query, k=k)

    return [doc.page_content for doc in results]
```

```python
import re

def retrieve_documents(query, k=5):

    query_upper = query.upper()

    # 1. Check for application numbers
    application_numbers = re.findall(r"\b\d{6}\b", query_upper)

    for appl_no in application_numbers:
        exact_results = vectorstore.get(
            where={"ApplNo": appl_no}
        )

        if exact_results["documents"]:
            return exact_results["documents"][:k]

    # 2. Check for exact drug names
    words = re.findall(r"\b[\w-]+\b", query_upper)

    for word in words:
        exact_results = vectorstore.get(
            where={"DrugName": word}
        )

        if exact_results["documents"]:
            return exact_results["documents"][:k]

    # 3. Fallback to semantic search
    results = vectorstore.similarity_search(query, k=k)

    return [doc.page_content for doc in results]
```

```python
def generate_answer(query):

    # Retrieve relevant FDA records
    retrieved_docs = retrieve_documents(query, k=3)

    # Combine retrieved records
    context = "\n\n".join(retrieved_docs)

    prompt = f"""
You are a helpful assistant for FDA drug information.

Answer the user's question using ONLY the information provided in the context below.

If the answer is not present in the context, say:
"I could not find that information in the provided FDA data."

Context:
{context}

User Question:
{query}

Answer clearly and concisely.
"""

    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )

    return response.text
```

```python
import re

def retrieve_documents(query, k=5):

    query_upper = query.upper()

    # 1. Application number
    application_numbers = re.findall(r"\b\d{6}\b", query_upper)

    for appl_no in application_numbers:
        results = rag_data[
            rag_data["ApplNo"].astype(str).str.upper() == appl_no
        ]

        if not results.empty:
            return results["document_text"].head(k).tolist()

    # 2. Exact drug name
    for drug in rag_data["DrugName"].dropna().unique():

        if str(drug).upper() in query_upper:

            results = rag_data[
                rag_data["DrugName"].astype(str).str.upper() == str(drug).upper()
            ]

            if not results.empty:
                return results["document_text"].head(k).tolist()

    # 3. Multi-condition search
    conditions = []

    if "NDA" in query_upper:
        conditions.append(
            rag_data["ApplType"].astype(str).str.upper() == "NDA"
        )

    # Sponsor
    for sponsor in rag_data["SponsorName"].dropna().unique():
        sponsor_upper = str(sponsor).upper()

        if sponsor_upper in query_upper:
            conditions.append(
                rag_data["SponsorName"].astype(str).str.upper() == sponsor_upper
            )
            break

    # Strength
    strength_matches = re.findall(r"\b\d+(?:\.\d+)?%", query_upper)

    if strength_matches:
        strength = strength_matches[0]

        conditions.append(
            rag_data["Strength"].astype(str).str.upper() == strength
        )

    # Apply conditions
    if conditions:

        mask = conditions[0]

        for condition in conditions[1:]:
            mask = mask & condition

        results = rag_data[mask]

        if not results.empty:
            return results["document_text"].head(k).tolist()

    # 4. Semantic fallback
    results = vectorstore.similarity_search(query, k=k)

    return [doc.page_content for doc in results]
```

```python
def generate_answer(query):

    # 1. Retrieve relevant FDA documents
    retrieved_docs = retrieve_documents(query, k=3)

    # 2. Combine retrieved documents
    context = "\n\n".join(retrieved_docs)

    # 3. Create prompt
    prompt = f"""
You are a helpful assistant for FDA drug information.

Answer the user's question using ONLY the information provided
in the context below.

If the answer is not present in the context, say:
"I could not find that information in the provided FDA data."

Context:
{context}

User Question:
{query}

Answer clearly and concisely.
"""

    # 4. Send prompt to Gemini
    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    return response.text
```
